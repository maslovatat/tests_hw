Часть 1. Two Sum двумя способами

**решение в лоб**

```python
class Solution:
    def twoSum(self, nums, target):
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]
        return []

```

Код работает ЦЕЛЫХ 1 770 миллисекунд, оказалось это очень медленно.
При этом, такое решение обходит по скорости лишь 20% участников, которые использовали Python

**Попробуем решить через dict**

```python
class Solution:
    def twoSum(self, nums, target):
        seen = {}
        for i, num in enumerate(nums):
            sum_num = target - num
            if sum_num in seen:
                return [seen[sum_num], i]
            seen[num] = i
        return []
```

image.png

Показатели точно улучшились

Решение «в лоб» имеет сложность O(n²), потому что перебирает все пары чисел двумя вложенными циклами. Решение через HashMap работает за O(n): массив проходится один раз, а поиск дополнения в словаре занимает в среднем O(1). На LeetCode решение «в лоб» показало Runtime 1733 ms (Beats 20.15%), а решение через HashMap — Runtime 3 ms (Beats 53.43%). Разница говорит о том, что при росте массива разрыв между O(n²) и O(n) увеличивается, а HashMap становится лучшим выбором для больших данных.

Скорректированный под требования LeetCode код:

```python
class Solution:
    def twoSum(self, nums, target):
        n = len(nums)
        for i in range(n):
            for j in range(i + 1, n):
                if nums[i] + nums[j] == target:
                    return [i, j]
        return []
```

Код проверяет каждую пару индексов i и j, где j > i, чтобы не использовать один элемент дважды.

Временная сложность:
O(n²) — внешний цикл выполняется до n раз, а внутренний в среднем тоже до
n раз.

Пространственная сложность:
O(1) — дополнительная память не зависит от размера списка

Решение корректно находит индексы, перебирая комбинации пока i < j. Оценка сложности O(n²) подтверждается на практике — 2180 ms делает решение одним из самых медленных (быстрее только 5.77% всех решений на Python)

**Решение агента с использованием HashMap**

```python
def two_sum_hashmap(nums, target):
    seen = {}  # значение -> индекс

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []  # пара не найдена


print(two_sum_hashmap([2, 7, 11, 15], 9))  # [0, 1]
```

Код снова немного неадаптирован для использования в LeetCode.

Скорректированный под требования LeetCode код:



```python
class Solution:
    def twoSum(self, nums, target):
        seen = {}
        for i, num in enumerate(nums):
            complement = target - num
            if complement in seen:
                return [seen[complement], i]
            seen[num] = i
        return []
```

Однако за эту скорость приходится платить памятью: словарь seen растёт пропорционально массиву (O(n)), и по памяти решение в нижней части рейтинга — 7.66%
